# **4. Split the dataset into a training set and testing set**

```python
#features
X = df.drop(columns=['Species']).values

#label
y = df['Species'].values
```

```python
X_train, X_test, y_train, y_test= train_test_split(X,y,test_size=0.2, random_state=5)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
print (X_train)
```

# **5. Create The Model**

**Choose machine learning algorithm**

```python
knn = KNeighborsClassifier(n_neighbors=13)
```

**Train the model**

```python
knn.fit(X_train, y_train)
```

**Test the model**

```python
X_pred = knn.predict(X_test)
metrics.accuracy_score(y_test, X_pred)
```
