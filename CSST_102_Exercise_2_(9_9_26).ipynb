# **1. Import Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn import metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

# **2.  Load and Understand Data Set**

```python
df = pd.read_csv('UCI_Wine_Dataset_178 (1).csv')
```

```python
df.head(5)
```

```python
df.drop(columns="Id", inplace = True)
```

```python
df.info()
```

```python
df.describe()
```

```python
df['class'].value_counts()
```

# **3. Visualize Data**

```python
plt.hist(df['Alcohol'])
plt.xlabel('Alcohol')
plt.ylabel('Number of rows')
plt.title('Histogram of Alcohol')
plt.show()
```

```python
plt.hist(df["Malicacid"])
plt.xlabel("Malicaced")
plt.ylabel("Id")
plt.show()
```

```python
colors = {1: 'red',
          2: 'blue',
          3: 'green'}
```

```python
plt.scatter(df['Alcohol'],
            df['Color_intensity'],
            c=df['class'].map(colors))

plt.xlabel("Alcohol")
plt.ylabel("Color_intensity")
plt.show()
```

# **4. Feaature Selection**

```python
from sklearn.feature_selection import f_classif

X = df.drop("class", axis = 1)
y = df["class"]
```

```python
F, p = f_classif(X, y)

feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "F_Score": F,
    "p_value": p
})

feature_scores = feature_scores.sort_values(
    "F_Score",
    ascending=False
)

feature_scores
```

```python
top_features = feature_scores.head(5)
```

```python
plt.figure(figsize=(10, 5))
plt.bar(top_features["Feature"], top_features["F_Score"])

plt.xticks(rotation=45)
plt.xlabel("Features")
plt.ylabel("F-Score")
plt.title("Top 5 Features")
plt.show()
```

# **5. Split the dataset: 80% training and 20% testing**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=5
)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# **6. Create The Model**

```python
knn = KNeighborsClassifier(n_neighbors=13)
```

Train Model

```python
knn.fit(X_train, y_train)
```

Test the Model

```python
y_pred = knn.predict(X_test)
```

Compute and display model accuracy

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print('Accuracy:', accuracy)
```

Display the confusion matrix

```python
metrics.confusion_matrix(y_test, y_pred)
```

# **7. Predict the classes of the test data**


```python
y_pred = knn.predict(X_test)
```

Accuracy

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print('Accuracy:', accuracy)
```

Precision

```python
precision = metrics.precision_score(y_test, y_pred, average='weighted')
print('\nPrecision:', precision)
```

Recall

```python
recall = metrics.recall_score(y_test, y_pred, average='weighted')
print('\nRecall:', recall)
```

F1-Score

```python
f1 = metrics.f1_score(y_test, y_pred, average='weighted')
print('\nF1-Score:', f1)
```
